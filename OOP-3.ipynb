```python
class App():
  bikes_list = []
  bikes_counter = 0
    
  def __init__(self):

    self.rev_list = []
    self.available_list = []
    self.in_charge_list = []   

  @staticmethod
  def total_counter():
    return App.bikes_counter
  
  def delete_bike(self,id):
    self.id = id
    for bike in App.bikes_list:
      if bike.id is self.id:
        App.bikes_list.delete(bike[id])
    App.bikes_counter -= 1
    return print(f"Bike with ID = {self.id} has been deleted")

  def check_revision(self):
    self.rev_list.clear()
    
    for bike in App.bikes_list:
      if bike.rev - bike.distance <= 100:
        self.rev_list.append(bike)
    
    print(f"Revision needed on bikes: {self.rev_list}")
    return self.rev_list
  
  def check_status(self):
      for bike in App.bikes_list:
        print(f"Bike ID : {bike.id} | Bike status:{bike.status}") 
        if bike.status == "available":
          self.available_list.append(bike)
        if bike.status == "in charge":
          self.in_charge_list.append(bike)
        else: pass
      
      if not self.available_list:
        print("Warning: There is no more available bikes on the network")

```

```python
class Bike(): 
  def __init__(self, id, status, rev = 2000):
    self.id = id
    self.rev = rev
    self.status = status
    App.bikes_counter += 1
    App.bikes_list.append(self)

class EBike(Bike):
  def __init__(self, id, status, bat, rev = 1000):

    super().__init__(id, status, rev)
    self.bat = bat
```

```python
bike_manager = App()

bike1 = Bike(1, "available")
bike2 = Bike(2, "in charge")
bike3 = EBike(3, "in charge",30)
bike4 = EBike(4, "in revision",100)
bike5 = EBike(5, "available",85)

```
